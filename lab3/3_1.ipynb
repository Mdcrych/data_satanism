```python
import pandas as pd

# Загружаем данные и очищаем пробелы в названиях стран
generation_df = (
    pd.read_csv('data/global-electricity-generation.csv')
    .assign(Country=lambda df: df['Country'].str.strip())
    .set_index('Country')
    .apply(pd.to_numeric, errors='coerce')
)

consumption_df = (
    pd.read_csv('data/global-electricity-consumption.csv')
    .assign(Country=lambda df: df['Country'].str.strip())
    .set_index('Country')
    .apply(pd.to_numeric, errors='coerce')
)

print('Производство электроэнергии:')
display(generation_df.head())
print('\nПотребление электроэнергии:')
display(consumption_df.head())
```

```python
# Суммарное потребление по годам
total_consumption_per_year = consumption_df.sum()
print('Суммарное потребление электроэнергии по годам (млрд. кВт*ч):')
print(total_consumption_per_year)
```

```python
# Максимальное производство одной страны за один год
max_val = generation_df.values.max()
print(f'Максимальное количество электроэнергии, произведенное одной страной за один год: {max_val:.2f} млрд. кВт*ч')
```

```python
# Страны с производством > 500 млрд. кВт*ч в среднем за последние 5 лет
avg_gen_last5 = generation_df.loc[:, '2017':'2021'].mean(axis=1)
countries_over_500 = avg_gen_last5[avg_gen_last5 > 500].index.tolist()
print('Страны, производящие в среднем более 500 млрд. кВт*ч за последние 5 лет:')
print(countries_over_500)
```

```python
# Топ 10% потребителей за последние 5 лет
avg_cons_last5 = consumption_df.loc[:, '2017':'2021'].mean(axis=1)
q90 = avg_cons_last5.quantile(0.9)
top10_consumers = avg_cons_last5[avg_cons_last5 >= q90].index.tolist()
print(f'Порог для топ 10% потребителей: {q90:.2f} млрд. кВт*ч')
print('10% стран, потребляющих больше всего электроэнергии в среднем за последние 5 лет:')
print(top10_consumers)
```

```python
# Страны, увеличившие производство более чем в 10 раз с 1992 по 2021
growth_ratio = generation_df['2021'] / generation_df['1992']
countries_10x = growth_ratio[growth_ratio > 10].index.tolist()
print('Страны, увеличившие производство более чем в 10 раз с 1992 по 2021 год:')
print(countries_10x)
```

```python
# Страны с суммарным потреблением > 100 млрд. кВт*ч и дефицитом производства
total_cons = consumption_df.sum(axis=1)
total_gen = generation_df.sum(axis=1)
deficit_mask = (total_cons > 100) & (total_gen < total_cons)
countries_deficit = total_cons[deficit_mask].index.tolist()
print('Страны с общим потреблением > 100 млрд. кВт*ч и дефицитом производства:')
print(countries_deficit)
```

```python
# Страна с наибольшим потреблением в 2020 году
country_max_2020 = consumption_df['2020'].idxmax()
max_val_2020 = consumption_df['2020'].max()
print(f'Страна, потратившая наибольшее количество электроэнергии в 2020 году: {country_max_2020} ({max_val_2020:.2f} млрд. кВт*ч)')
```

```python
# 10% стран с наименьшей средней производительностью за первые 5 лет (1992-1996)
avg_gen_first5 = generation_df.iloc[:, :5].mean(axis=1)
q10 = avg_gen_first5.quantile(0.1)
bottom10 = avg_gen_first5[avg_gen_first5 <= q10].index.tolist()
print(f'Порог для 10% стран с наименьшей производительностью: {q10:.4f} млрд. кВт*ч')
print(f'\n10% стран с наименьшей производительностью за первые 5 лет (1992-1996):')
for country in bottom10:
    print(f'  - {country}')
```
